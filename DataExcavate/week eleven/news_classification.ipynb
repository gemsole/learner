```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB # 一种专门用于文本分类的算法（朴素贝叶斯）
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report

# 1. 从本地加载数据集 并只用4个类别
categories = ['alt.atheism', 'soc.religion.christian', 'comp.graphics', 'sci.med']

print("正在从本地加载数据...")
try:
    # 核心参数：download_if_missing=False。只使用本地文件。
    # 核心参数：remove=('headers', 'footers', 'quotes')。去除邮件头、尾部和引用，纯化文本内容，提高准确率。
    newsgroups_train = fetch_20newsgroups(
        # data_home=r'D:\Dane\Documents\Trae\DataExcavate\week eleven',
        data_home=r'./data',
        # 指定数据集目录
        subset='train', 
        categories=categories,
        download_if_missing=False, 
        remove=('headers', 'footers', 'quotes')
    )
    newsgroups_test = fetch_20newsgroups(
        # data_home=r'D:\Dane\Documents\Trae\DataExcavate\week eleven',
        data_home=r'./data',
        # 指定数据集目录
        subset='test', 
        categories=categories,
        download_if_missing=False,
        remove=('headers', 'footers', 'quotes')
    )
    print(f"成功加载本地数据！训练集样本数: {len(newsgroups_train.data)}，测试集样本数: {len(newsgroups_test.data)}")
    
except FileNotFoundError:
    print("找不到本地数据文件！请确保 20news-bydate.tar.gz 已正确放置在 C:\\Users\\xxx(替换成自己的电脑名称)\\scikit_learn_data\\20news_home\\ 目录下。")
    exit()



# 2. 构建机器学习管道并训练模型
print("\n正在提取特征并训练模型...")
# Pipeline: 先将文本转为 TF-IDF 矩阵，再送入朴素贝叶斯分类器
model = make_pipeline(
    TfidfVectorizer(max_features=5000), # 保留最重要的5000个词汇
    MultinomialNB()                     # 多项式朴素贝叶斯
)

# 开始训练
model.fit(newsgroups_train.data, newsgroups_train.target)


# 3. 在测试集上评估模型
print("\n在测试集上进行预测评估...")
predicted = model.predict(newsgroups_test.data)

print("\n=== 模型分类评估报告 ===")
print(classification_report(
    newsgroups_test.target, 
    predicted, 
    target_names=newsgroups_test.target_names
))


# 4. 实际预测
print("===实际预测===")

# 使用医疗相关的英文文本（模拟 sci.med 类别）
new_text = """
The patient was admitted with severe symptoms. 
Doctors performed a series of blood tests and an MRI scan to check for any abnormalities. 
The prescribed medication will be taken twice a day for two weeks.
"""

# 进行预测
predicted_class_index = model.predict([new_text])[0]
predicted_category = newsgroups_test.target_names[predicted_class_index]

print(f"待预测文本:\n{new_text.strip()}\n")
print(f"模型预测分类结果: 【{predicted_category}】")
```
