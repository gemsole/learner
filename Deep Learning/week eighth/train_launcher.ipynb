# 训练启动器

Cell 1: 环境检查并后台启动 `demo.py` 训练（仅在有 GPU 时启动）

Cell 2: 轮询查看 `/root/train.log` 训练进度

```python
import torch, sys, subprocess
print(sys.executable)
print("torch", torch.__version__)
cuda = torch.cuda.is_available()
print("cuda_available", cuda)
if cuda:
    print("device", torch.cuda.get_device_name(0))
    out = open("/root/stdout.log", "w")
    p = subprocess.Popen([sys.executable, "/root/demo.py"], cwd="/root", stdout=out, stderr=subprocess.STDOUT, start_new_session=True)
    print("started pid", p.pid)
else:
    print("NOT starting - no GPU visible")
```

```python
import os
if os.path.exists("/root/train.log"):
    print(open("/root/train.log").read())
else:
    print("train.log not created yet")
if os.path.exists("/root/stdout.log"):
    print("--- stdout tail ---")
    print("".join(open("/root/stdout.log").readlines()[-20:]))
```
